```python
import pandas as pd
path = '/staging/users/tpadhi1/agent-jail-breaking/datasets/pinxian-pipeline/original/original_april16/instagram/type5_v12u_output.csv'
df = pd.read_csv(path)
```

```python
df['csv2_input'].value_counts()
```

```python
import os

# ok so for each of the df['csv2_input'].value_counts(), 200 for each category, I want a csv, with just 10 conversations per input, write the pandas dataframe to csv

output_dir = 'output_csvs'
os.makedirs(output_dir, exist_ok=True)

# create a empty df
output_df = pd.DataFrame()
for category, count in df['csv2_input'].value_counts().items():
    # Filter the DataFrame for the current category
    category_df = df[df['csv2_input'] == category]
    
    # Take the first 10 conversations (or fewer if less than 10)
    sample_df = category_df.head(2)
    output_df = pd.concat([output_df, sample_df], ignore_index=True)

```

```python
output_df['csv2_input'].value_counts()

# create categories from the csv2_input column
# take the category before the ( 
output_df['category'] = output_df['csv2_input'].str.split('(').str[0].str.strip()
```

```python
output_df.head()
```

```python
# create a list of tuples from (category, conversation) from the df

category_conversation_list = []
for index, row in output_df.iterrows():
    category = row['category']
    conversation = row['agent3_output_converted']
    category_conversation_list.append((category, conversation)) 

```
